### Quanto deve medir x para que a calha tenha capacidade máxima ?



### A capacidade de escoamento de água da calha é, formalmente, a vazão! 

###  <center> $ Q(A, v) = A \cdot v $ </center>

#### Q(A,v) é a vazão (cm3/s);
#### A é a área da seção (cm2); e
#### v é a velocidade do fluído (cm/s).

### Supondo v constante, a vazão torna-se diretamente proporcional à área da seção! Portanto, maximizando A implica em maximizar Q(A,v)

```python
#CÉLULA GRADIENTE-LIB-01
import numpy as np
import math
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
```

## 1 - Qual é a função matemática que define a área da seção da calha?

```python
#CÉLULA GRADIENTE-LIB-02

def area(x) :
    return 30*x-2*x**2
```

## 2 - Plote o gráfico da função no intervalo [0,15]

```python
#CÉLULA GRADIENTE-LIB-03

x = np.arange(0,15,0.001)
y = area(x)

plt.figure(figsize=(8,8))
plt.grid()
plt.plot(x, y)
```

## 3 - Implemente o algoritmo de gradiente.

```python
#CÉLULA GRADIENTE-LIB-04
# x(i+1) = x(i) + t*(30-4x(i))
# residuo = | x(i+1) - x(i) |. Processo iterativo cessa quando resíduo é menor que “precisão desejada”

t = 0.4
epsilon = 0.0001

i = 0
vx = []
vx.append(-3) #(np.random.rand() * 15)

error = 1 + epsilon
vError = []
vError.append(error)

while ( error > epsilon ) :    
    newX = vx[i] + t*(30-4*vx[i])
    vx.append(newX)
    error = abs(vx[i+1] - vx[i])
    vError.append(error)
    i += 1   
    
print(i)
print(error)
```

## 4 - Plote o gráfico (vx X vError)

```python
#CÉLULA GRADIENTE-LIB-05
plt.figure()
plt.grid()
plt.plot(vx[1:], vError[1:])
```
