```python
def generateMetadata(df):
    data = []
    for f in df.columns:
        # definindo o uso (entre rótulo, id e atributos)
        if f == 'RainTomorrow':
            role = 'RainTomorrow' # rótulo
        else:
            role = 'input' # atributos

        # definindo o tipo do dado
        if df[f].dtype == object:
            level = 'nominal'
        elif df[f].dtype == float or df[f].dtype == np.float64:
            level = 'interval'
      
        keep = True

        # cria o tipo de dado
        dtype = df[f].dtype

        # cria dicionário de metadados
        f_dict = {
            'varname': f,
            'role': role,
            'level': level,
            'keep': keep,
            'dtype': dtype
        }
        data.append(f_dict)

    meta = pd.DataFrame(data, columns=['varname', 'role', 'level', 'keep', 'dtype'])
    meta.set_index('varname', inplace=True)
    
    return meta
```

```python
meta = generateMetadata(weather)
display(meta)
```
