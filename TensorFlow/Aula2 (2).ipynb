```python
naturais1 = np.arange(1,101)
naturais1
```

```python
naturais2 = np.linspace(1,100)#números igualmente espaçados
naturais2
```

```python
naturais = tf.constant(list(range(1,101)))
naturais
```

```python
soma = tf.reduce_sum(naturais)
soma
```

```python
with tf.Session() as s:
    print(s.run(soma))
```

```python
np.logspace(1,100,num=100,base=3)#números igualmente espaçados em escala logarítmica
```

```python
termos = tf.constant(np.logspace(1,100,num=100,base=3))
termos
```

```python
soma = tf.reduce_sum(termos)
soma
```

```python
with tf.Session() as s:
    print(s.run(soma))
```
