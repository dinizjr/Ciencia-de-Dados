```python
import pandas as pd
import tensorflow as tf
%matplotlib inline
from sklearn.model_selection import train_test_split
```

```python
base = pd.read_csv('breast-cancer.csv',encoding='UTF-8')
base.head()
```

```python
base['class'].unique()
```

```python
base.shape
```

```python
def convert_class(rotulo):
    if rotulo == 'benign':
        return 1
    else:
        return 0
```

```python
base['bare_nuclei'].unique()
```

```python
base['bare_nuclei'].replace('?', '0' ,inplace=True)
base['bare_nuclei'].unique()
```

```python
base['class'] = base['class'].apply(convert_class)
```

```python
base.head()
```

```python
X = base.drop(['class','code_number'], axis = 1)
X.head()
```

```python
y = base['class']
y.head()
```

```python
X.columns
```

```python
X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(X, y, test_size = 0.3)
```

```python
X_treinamento.shape
```

```python
X_teste.shape
```

```python
nome_colunas_numericas = ['clump_thickness', 'uniformity_cell_size', 'uniformity_cell_shape', 'marginal_adhesion', 'single_epitelial_cell_size','bare_nuclei','bland_chromation','normal_nucleoli','mitoses']
nome_colunas_numericas
```

```python
colunas_numericas = [ tf.feature_column.numeric_column(key = c) for c in nome_colunas_numericas ]
colunas_numericas
```

```python
funcao_treinamento = tf.estimator.inputs.pandas_input_fn(x = X_treinamento, y = y_treinamento,
                                                        batch_size = 32, num_epochs = None, shuffle = True)
```

```python
classificador = tf.estimator.LinearClassifier(feature_columns=colunas_numericas)
```

```python
classificador.train(input_fn=funcao_treinamento, steps = 10000)
```

```python
funcao_previsao = tf.estimator.inputs.pandas_input_fn(x = X_teste, batch_size = 32, shuffle = False)
```

```python
previsoes = classificador.predict(input_fn=funcao_previsao)
```
