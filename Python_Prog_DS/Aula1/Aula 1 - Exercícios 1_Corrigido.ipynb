## 7 - Dada o vetor [1, 3, 0, 3, 0, 3, 2, 0, 1, 0], descubra os índices dos elementos diferentes de zero.

````python
arr = np.array([1,3,0,3,0,3,2,0,1,0])
np.nonzero(arr)
````

```
array([0, 1, 3, 5, 6, 8], dtype=int64)
```
