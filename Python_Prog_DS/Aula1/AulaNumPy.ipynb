# Principais métodos da NumPy

## Método: arange()

Semelhante ao range do Python, porém devolve um array com os elementos que estão dentro do intervalo selecionado.

```python
np.arange(0, 10)
```

```python
np.arange(0, 10, 2)
```

```python
np.arange(0, 101, 10)
```

```python
np.arange(-10, 1)
```

## Método zeros()

Retorna uma matriz com a dimensão informada, preenchida com zeros.

```python
np.zeros(3)
```

```python
np.zeros(4, dtype=int)
```

```python
np.zeros((3, 3))
```

```python
np.zeros((4, 2), dtype=int)
```

```python
np.zeros([3, 3, 2])
```

## Método: ones()

Da mesma forma do método zeros() retorna uma matriz com a dimensão informada, preenchida com 1.

```python
np.ones(3)
```

```python
np.ones(3, dtype=int)
```

```python
np.ones((3, 4))
```

## Método: linspace()

Cria uma sequencia de números igualmente espaçados entre os limites dados.

```python
np.linspace(0, 50, 5, dtype=int)
```

```python
np.linspace(0, 100, 10, dtype=int)
```

# Sub-biblioteca random

## Método: rand()

Retorna uma matriz com a dimensão informada, preenchida com números aleatórios entre 0 e 1, seguindo uma distribuição uniforme.

```python
np.random.rand(5)
```

```python
np.random.rand(3, 4)
```

É possível incluir operações na criação de números aleatórios

```python
np.random.rand(5) * 100 
```

```python
np.random.rand(5) ** 2
```

```python
np.random.rand(2, 2) * 10
```

## Método: randn()

Retorna uma matriz com a dimensão informada, preenchida com números aleatórios, seguindo uma distribuição normal.

```python
np.random.randn(5)
```

```python
np.random.randn(3, 4) * 10
```

## Método: randint()

Retorna um número inteiro aleatório entre o intervalo informado ou uma matriz unidimensional, preenchida com números aleatórios inteiros, entre o intervalo informado.

```python
np.random.randint(1, 10, 15)
```

```python
np.random.randint(1, 1000, 5)
```

```python
np.random.randint(1, 1000, 5) * 10
```

# Métodos e atributos do objeto NumPy Array

## Método: reshape()

Reorganiza um array em uma dimensão diferente a original

```python
arr = np.random.rand(10)
```

```python
arr
```

```python
arr.reshape(2, 5)
```

```python
arr
```

```python
arr.reshape(5, 2).reshape(2, 5)
```

```python
arr_int = np.random.randint(0, 100, 50)
```

```python
arr_int
```

```python
arr_int.reshape(5, 10)
```
