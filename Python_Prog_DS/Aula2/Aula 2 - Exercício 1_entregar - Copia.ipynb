### Exercício 7 - Dado o array [4, 95, 37, 64, 42, 19, 38, 46, 83, 48, 67, 98, 21, 10, 88,55], obtenha todos os itens que estão entre 35 e 55.

````python
arr = [4,95,37,64,42,19,38,46,83,48,67,98,21,10,88,55]
arr2 = np.sort(arr)
arr2
````

````python
#mask1 = (arr2 > 35)
arr2[np.where(np.logical_and(arr2 > 35, arr2 < 56))]
````
