* lat: Variável String, Latitude

* twp: Variável String, Township
* addr: Variável String, Endereço
* e: Variável String, variável Dummy (sempre 1)



### Leia o arquivo csv (911.csv) como um dataframe chamado _calls_

```python
import pandas as pd
import numpy as np
```

### -_Ler o arquivo_

```python
calls = pd.read_csv('911.csv')
calls.head()
calls
```

### Quantas linhas e colunas tem o arquivo? Qual a quantidade de memória necessária para carregá-lo?

### -_Mostrar quantidade de linhas e colunas,tipos das colunas e memória usada pelo arquivo_

```python
calls.info()
```

### Quais são os top 5 CEPs nas chamadas 911?

### -_Mostrar os 5 CEPs com mais chamados_

```python
calls.zip.value_counts().nlargest(5)
```

### Quais são os 5 principais municípios nas chamadas 911?

### -_Mostrar os 5 municípios com mais chamados_

```python
calls.twp.value_counts().nlargest()
```

### Quantos códigos de título únicos existem?

### -_Contar quantos códigos de títulos únicos há no arquivo_

```python
calls.title.value_counts().count()
```

### Na coluna _title_ existem Razões / Departamentos especificados antes do código do título. Estes são _"EMS"_, _"Fire"_ e _"Traffic"_. Crie uma nova coluna chamada _Reason_ que conterá esse valor em cada linha

#### Por exemplo, se o valor da coluna do título for EMS: BACK PAINS / BLESSOR, o valor da coluna Reason será EMS.

### -_Criação da coluna Reason separando as Razões/Departamentos da coluna title_

```python
arr = np.array(calls.loc[:,'title'])
calls['Reason'] = [f.split(sep=':')[0] for f in arr]
calls

```

### Qual é a razão mais comum para uma chamada do 911 com base nessa nova coluna?

### -_Mostrar a razão mais comum da coluna Reason_

```python
calls.Reason.value_counts().nlargest(1)
```

### Qual é o tipo de dados dos objetos na coluna _timeStamp_?

### -_Mostrar o tipo de dados da coluna timeStamp_

```python
calls.timeStamp.dtype
```

### Converta a coluna _timeStamp_ para DateTime. Use [pd.to_datetime](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.to_datetime.html) para essa operação.

### -_Conversão da coluna timeStamp para DateTime_

```python
calls.timeStamp = pd.to_datetime(calls.timeStamp)
calls.timeStamp.dtype
```

### Crie 3 novas colunas chamadas _Hour_, _Month_ e _Day of Week_. Você criará essas colunas com base na coluna timeStamp

### -_Separação de horas, meses e dias da semana da coluna timestamp e criado colunas novas_

```python
calls['Hour'] = calls['timeStamp'].dt.hour
calls['Month'] = calls['timeStamp'].dt.month
calls['Day_of_Week'] = calls['timeStamp'].dt.dayofweek
calls
```

### Observe como o dia da demana é um número inteiro de 0-6. Mapeie os nomes das seqüências reais para o dia da semana, nessa mesma coluna:

- 0 => Mon
- 1 => Tue
- 2 => Wed
- 3 => Thu
- 4 => Fri
- 5 => Sat
- 6 => Sun

### -_Conversão dos dias da semana de número para nome do dia_

```python
calls['Day_of_Week'] = calls.timeStamp.dt.strftime('%a')
calls
```

### Qual a quantidade de chamadas de cada motivo (_Reason_) por mês?

### -_Agrupar chamados por motivo e por mês_

```python
calls.groupby(['Reason','Month']).size()
```

### Qual foi o dia com a maior quantidade de chamadas? Quantas chamadas foram executadas neste dia?

### -_Mostrar o dia com mais chamados e quantidade de chamados neste dia_

```python
calls.timeStamp.dt.strftime('%Y-%m-%d').value_counts().nlargest(1)
```

### Levando-se em consideração apenas atendimentos feitos às sextas-feiras, qual a hora do dia com o maior número de chamadas?

### -_Mostrar a hora com mais chamados nas sextas-feiras_

```python
calls[calls.Day_of_Week=='Fri'].Hour.value_counts().nlargest(1)
```

### Levando-se em consideração os atendimentos relacionados a incêndio (_Reason_ = _Fire_) qual a quantidade de chamados aos sábados?

### -_Mostrar a quantidade de chamados de incêndio aos sábados_

```python
calls[(calls.Reason=='Fire') & (calls.Day_of_Week=='Sat')].Day_of_Week.value_counts().nlargest(1)
```
