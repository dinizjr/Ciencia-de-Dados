### Exercício 4 - Dado um array de 2 dimensões, como trocar 2 colunas entre si?

## Inversão das colunas 2 e 3:
### Foi criado uma cópia do array original,
### inserido a coluna 3 da cópia no array original,
### e inserido a coluna 2 da cópia no array original.

````python
arr = np.arange(0,25).reshape(5,5)
arr_copy = np.copy(arr)
arr[:,2] = arr_copy[:,3]
arr[:,3] = arr_copy[:,2]
arr
````

### Exercício 6 - Dado o array [1, 2, 3] criar o array [[1, 1, 1], [2, 2, 2], [3, 3, 3]]

## Repetição do array [1,2,3] em colunas:
### Foi repetido 3 vezes o array [1,2,3] em um array de uma dimensão usando o tile,
### depois foi transformado em um array 3 por 3 com o reshape,
### e finalmente invertido linhas por colunas com rollaxis.

````python
arr = [1,2,3]
arr = np.rollaxis(np.tile([1,2,3],3).reshape(3,3),1)
arr
````

**Resposta:**

```
array([[1, 1, 1],
       [2, 2, 2],
       [3, 3, 3]])
```

### Exercício 7 - Dado o array [4, 95, 37, 64, 42, 19, 38, 46, 83, 48, 67, 98, 21, 10, 88], obtenha todos os itens que estão entre 35 e 55.

## Seleção dos valores entre 35 e 55 do array especificado:
### O array foi ordenado e em seguida foi usado where para filtrar os valores solicitados.

````python
arr = [4,95,37,64,42,19,38,46,83,48,67,98,21,10,88,55]
arr2 = np.sort(arr)
arr2[np.where(np.logical_and(arr2 > 35, arr2 < 56))]
````

**Resposta:**

```
array([37, 42, 38, 46, 48, 55])
```
