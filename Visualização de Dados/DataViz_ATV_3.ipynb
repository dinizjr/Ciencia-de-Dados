```python
%matplotlib inline
import numpy as np
import pandas as pd
import plotly.plotly as py
import plotly.graph_objs as go
import matplotlib.pyplot as plt
import plotly.figure_factory as ff

from plotly import tools
from plotly.offline import download_plotlyjs, init_notebook_mode, plot, iplot

init_notebook_mode(connected=True)

#import plotly
import plotly.offline as py

from IPython.core.pylabtools import figsize
```

```python
url ="https://raw.githubusercontent.com/matheusmota/dataviz2018/master/resources/datasets/cars.csv"
cars = pd.read_csv(url, sep=',')
cars.columns = cars.columns.str.replace(' ', '_')
cars.head()
#cars.info()
```

```python
marcas = pd.DataFrame(cars.Make.value_counts().reset_index(name='qtd').rename(columns={'index':'marcas'}))

trace = go.Scatter(x=marcas['marcas'],y=marcas['qtd'],mode = 'lines+markers')
data = [trace]
layout = go.Layout(
    title='Quantidade de veículos por marca',
    xaxis=dict(title='Marcas',titlefont=dict(size=20)),
    yaxis=dict(title='Quantidade',titlefont=dict(size=20)))
py.iplot(go.Figure(data = [trace], layout=layout))
```

```python
fuel = pd.DataFrame(cars.Classification.value_counts().reset_index(name='qtd').rename(columns={'index':'fuel'}))
#a = fuel.sort_values(by='qtd',ascending=False)
trace = go.Bar(x=fuel['fuel'],y=fuel['qtd'])
data = [trace]
layout = go.Layout(
    title='Quantidade de veículos por combústivel)',
    xaxis=dict(title='Transmissão',titlefont=dict(size=20)),
    yaxis=dict(title='Quantidade',titlefont=dict(size=20)))
py.iplot(go.Figure(data = [trace], layout=layout))
```

```python
#df['service'] = [words.split(sep=':')[0] for words in np.array(df.loc[:,'title'])]
cars['Cylinder'] = [words.split(sep='L')[0] for words in np.array(cars.loc[:,'Engine_Type'])]
cars.Cylinder.head()

```

```python
marcas = pd.DataFrame(cars.Make.value_counts().reset_index(name='qtd').rename(columns={'index':'marcas'}))

trace = go.Bar(x=marcas['marcas'],y=marcas['qtd'])
data = [trace]
layout = go.Layout(
    title='Quantidade de veículos por marca)',
    xaxis=dict(title='Marcas',titlefont=dict(size=20)),
    yaxis=dict(title='Quantidade',titlefont=dict(size=20)))
py.iplot(go.Figure(data = [trace], layout=layout))
```

```python
trace_women = go.Bar(x=cars.Make,
                  y=cars.Highway_mpg,
                  name='Highway_mpg',
                  marker=dict(color='#ffcdd2'))

trace_gap = go.Bar(x=cars.Make,
                y=cars.Horsepower,
                name='Cavalos',
                marker=dict(color='#59606D'))

data = [trace_women,trace_gap]

layout = go.Layout(title="Average Earnings for Graduates",
                   xaxis=dict(title='School'),
                   yaxis=dict(title='Salary (in thousands)'))

fig = go.Figure(data=data, layout=layout)

py.iplot(fig,)
```

```python
trace = go.Bar(x=cars.Make,y=cars.Highway_mpg)
data = [trace]
layout = go.Layout(
    title='Consumo por marca',
    xaxis=dict(title='Marcas',titlefont=dict(size=20)),
    yaxis=dict(title='Quantidade',titlefont=dict(size=20)))
py.iplot(go.Figure(data = [trace], layout=layout))
```
