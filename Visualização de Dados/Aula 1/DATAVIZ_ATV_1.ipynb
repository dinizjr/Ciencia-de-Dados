```python
ax = a.plot(kind='bar',rot=90, figsize=(10,7),color='coral',fontsize=13,label='Categoria')
#a.set_alpha(0.8)
ax.set_title('Porcentagem de acidentes por categoria', fontsize=18)
ax.set_ylabel('Número de Acidentes', fontsize=18)
ax.set_yticks([0,400,800,1200,1600,2000,2400])
ax.set_ylim((0,2500))
totals = []
for i in ax.patches:
    totals.append(i.get_height())
total = sum(totals)
for i in ax.patches:
    ax.text(i.get_x(), i.get_height()+25, \
          str(round((i.get_height()/total)*100, 2))+'%', fontsize=15,
             color='dimgrey')
plt.style.use('ggplot')
plt.legend()
plt.show()
```
