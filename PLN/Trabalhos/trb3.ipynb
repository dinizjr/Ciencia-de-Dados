```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv('SW_EpisodeV.txt', sep='\t')
df.head(30)
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
df.isna().sum()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(df)
print(tfidf_matrix.shape)
```

```python
def getIndexSimilarity(question):
    query_vect = tfidf_vectorizer.transform([question])
    similarity = cosine_similarity(query_vect, tfidf_matrix)
    i = np.where(similarity[0] == similarity.max())
    return i[0][0]
```

```python
qst = input('The question is: ')
#qst = Who killed lincoln? #Did Lincoln beat in John
index = getIndexSimilarity(qst)
print('The most similar question was: {} \nThe answer is: {}'.format(df.iloc[index],df.iloc[index]))
```
