<h2>1. Resolução de Coreferência</h2>

<p>A coreferência ocorre quando há alguma expressão referenciando a outra entidade presente no texto (chamada de referente). Perceba que pode haver mais de uma referência para o mesmo referente. Note também que o referente pode ser uma entidade nomeada ou não. Vejamos o exemplo a seguir.</p>

<p>"João da Silva foi eleito o empresário do ano. Ele fundou a maior rede de restaurante da cidade. Por isso, foi destaque na revista mais famosa da cidade, a Revista Foco."</p>

<p>Nesse exemplo, temos os seguintes tipos de coreferência:</p>

* **Anáfora**: Ocorre quando o referente vem antes da referência (chamada de anáfora), como em Ele -> João da Silva. Como se trata de um pronome, diz-se que trata-se de uma anáfora pronominal

* **Referência Nominal**: Quando um sintagma nominal se refere a uma entidade previamente mencionada no texto, como em "empresário do ano" -> "João da Silva"

* **Catáfora**: Esse fenômeno ocorre quando a referência aparece antes do referente no texto, como em "revista mais famosa da cidade" -> "Revista Foco"

<p>A resolução de coreferência consiste em encontrar as referências para entidades presentes no texto. Trata-se de uma tarefa essencial para diversos sistemas de Processamento de Linguagem Natural, commo sumarização automática ou sistemas diversos para extração de informação.</p>

<p>Porém, o grande desafio para resolução de coreferências surge quando há mais de um candidato a referente que concorda (em gênero e número) com a anáfora (ou catáfora). Exemplo:</p>

<p>"João emprestou uma ferramenta de Paulo, e ele a quebrou".</p>



<h2>2. Técnicas para Resolução de Coreferência </h2>

<p>Diversas abordagens com algorítmos baseados em teorias linguísticas e aprendizado de máquina já foram propostas para tratar o problema da resolução de coreferência. Nas pesquisas mais tradicionais, utilizam-se de modelos de aprendizado de máquina que contam com atributos como:
</p>

* Distância (em palavras) entre entidade e candidato a anáfora;
* Classificação sintática das duas palavras (sujeito, objeto direto, objeto indireto);
* POS tag das duas palavras;
* Indicador se o referente se trata de um substantivo composto;
* Indicador se o referente se trata de um substantivo coletivo;

<p>Porém, abordagens mais contemporâneas como a implementada na biblioteca neuralcoref do Spacy (https://github.com/huggingface/neuralcoref) utilizam-se de word embeddings (Word2Vec) das palavras próximas, além de outros atributos como a distância entre palavras, para criar um classificador utilizando redes neurais. No caso desse modelo, conforme exibido na imagem à seguir, duas redes foram treinadas: uma para indicar se há algum referente para um candidato a anáfora, e outra para classificar se a referência entre um candidato a anáfora e um referente é válida.</p>

```python
import spacy
nlp = spacy.load('en')

import neuralcoref
neuralcoref.add_to_pipe(nlp)

doc = nlp(u'My sister has a dog. She loves him.')

doc._.has_coref
doc._.coref_clusters
```

```python
doc = nlp(u'My sister has a dog. She loves him but he is very silly. Once a cat crossed his way and he started fleeing in fear.')

doc._.has_coref
doc._.coref_clusters
```

```python
doc = nlp(u'My sister has a dog. She loves him but he is very silly. \
Once a cat crossed his way and he started fleeing in fear. At the end, the poor kitty was just trying to find some food nearby.')

doc._.has_coref
doc._.coref_clusters
```

<p>Agora vamos utilizar o Spacy para listar as entidades e suas coreferências.</p> 

```python
doc = nlp(u'Mark Johnson has recently announced he will be attending to patients at St. Paul hospital from May 3rd. \
The hospital team stated they are very honnored on having him -- a renowned cardiologist -- as part of their staff.')

doc._.has_coref
doc._.coref_clusters
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
    print(ent._.coref_cluster)
```
