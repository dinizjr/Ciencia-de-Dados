### 1 -  Instruções - Treinamento

1. Faça a configuração dos parâmetros de treinamento (eta, epsilon, numero de entradas, camada oculta e saida)
2. Faça a carga do arquivo de treinamento com o método 
        model.loadDataFile('ds_resonance_training.csv')
3. Faça a separação entre os dados de entrada {'x1', 'x2', 'x3'} e saida {'d'}
4. Execute o treinamento em batelada a partir do método 
        model.mlp_train_batch(eta, epsilon, vInputsTrain.T, vOutputsTrain.T, dictLayers, max_epochs) 
5. Execute o treinamento online a partir do método 
        model.mlp_train_online(eta, epsilon, vInputsTrain.T, vOutputsTrain.T, dictLayers, max_epochs) 

```python
#CÉLULA MLP-MAIN-01

#Criação da tabela 1
table_MSE = pd.DataFrame(columns=['Erro Quadrático Médio', 'Número de Épocas' ])

#Parâmetros
#1 - Configuração dos parametros de treinamento
#IMPLEMENTE O CÓDIGO AQUI
eta        = 0.1
epsilon    = 0.000001
nInputs    = 3
nHidden    = 10
nOutputs   = 1
max_epochs = 1000
dictLayers = {'nLayers':2, 'vLayers':np.array([nInputs, nHidden, nOutputs])}

#2 - Carga dos dados de treinamento
dfTrainingFile = model.loadDataFile('ds_resonance_training.csv')

nSamples = len (dfTrainingFile['x1'])

#3 - Separação dos dados de treinamento em entradas e saida
vInputsTrain  = np.array(dfTrainingFile[['x1', 'x2', 'x3']])
vOutputsTrain = np.array(dfTrainingFile['d'].values.reshape(nSamples,1))

#4 - Execução do treinamento

epocas = []
list_out = []
for train in range(5):
    list_MSE = []
    
    #for t in range(0,1):
    (W_1, W_2, B_1, B_2, MSE) = model.mlp_train_online(eta, epsilon, vInputsTrain.T, vOutputsTrain.T, dictLayers, max_epochs) 
        
    #print(W_1)
    #print('-----')
    #print(W_2)
    #print('-----')
    
    #popular lista com o número de épocas
    epocas.append(len(MSE)-1)
    
    #popular lista de dicionários com as saídas
    list_out.append({'Treino':train+1,'Época':len(MSE)-1,'MSE':MSE,'W_1':W_1,'W_2':W_2,'B_1':B_1,'B_2':B_2})
    
    #print(len(MSE)-1)
    #print(MSE[-1:])
    
    #popular lista com os erros e épocas
    list_MSE.extend(MSE[-1:])#último erro
    list_MSE.append(len(MSE)-1)#épocas
    #print(list_MSE)
    
    #Calcula o Erro Quadratico Medio
    #MSE_temp = model.calcMSE(vInputsTrain, vOutputsTrain, [W_1, W_2], [B_1, B_2])
    #list.append( MSE_temp )
    
    #popular tabela 1
    table_MSE.loc['T'+str(train+1),:] = list_MSE
    
    #table_MSE.iloc[train,:] = list_MSE
    
#print(list_out)
#print(table_MSE)
#print(epocas)
#print(my_set)

#Verificação dos 2 treinos com maiores épocas e plot de seus gráficos de erro
bkp_plot = []
for k in range(2):
    for i in range(5):
        if list_out[i]['Época'] == max(epocas) and i not in bkp_plot:  
            print('-----------------------------------------------')
            print('################# GRÁFICO {} ###################'.format(k+1))
            print(' Erro quadrático médio(MSE) X número de épocas')
            print('-----------------------------------------------')
            print( '-------- Treinamento {} * {} Épocas ------------'.format(list_out[i]['Treino'],list_out[i]['Época']))
            plt.figure()
            plt.plot(range(len(list_out[i]['MSE'])),list_out[i]['MSE'])
            plt.show()
            bkp_plot.append(i)
            break
    epocas.remove(max(epocas))
print('-----------------------------------------------')
print('################# TABELA 1 ####################')
print('      MSE e nº de épocas por treinamento')
print('-----------------------------------------------')
table_MSE
```

## Baseado na tabela do item 2, explique de forma detalhada por que tanto o erro quadrático médio quanto o número de épocas variam de treinamento para treinamento.

### Porque com o backpropagation, conforme vai corrigindo os pesos o erro vai diminuindo, e como os pesos iniciais sempre são diferentes entre as épocas, o número de épocas varia.

### 2 -  Instruções - Verificação de resultados

1. Execute a operação do Perceptron para os dados de treinamento a partir do método
        y_pred = model.mlp_operation(vInputsTrain.T, [W_1, W_2], [B_1, B_2])
2. Plote o gráfico dos dados de previsão da rede, y_pred
3. Plote o gráfico dos dados reais, vOutputs
4. Calcule o erro relativo medio

<center> $ERM = avg(E\_ABS) $ </center>

onde E_ABS é o erro absoluto das amostras, ou seja

        E_ABS = np.abs(y_pred - vOutputsTrain)
        ERM   = np.average(E_ABS)
    
5. A variancia pode ser calculada a partir do erro absoluto

        var_ERM = np.var(E_ABS)

```python
#1 - execução do MLP para dados de treino
#IMPLEMENTE O CÓDIGO AQUI
y_pred = model.mlp_operation(vInputsTrain.T, [W_1, W_2], [B_1, B_2])

#2,3 - Graficos da resposta prevista versos dados esperados (reais)
plt.figure(figsize=(12,8))

#Dados de predição
y_pred = y_pred.T

#IMPLEMENTE O CÓDIGO AQUI - Grafico (y_pred)
plt.plot(range(len(y_pred)), y_pred, 'ro-')
#Dados reais
#IMPLEMENTE O CÓDIGO AQUI - Grafico (vOutputs)
plt.plot(range(len(vOutputsTrain)), vOutputsTrain, 'b+-')
#Legenda
plt.gca().legend(('Previsão', 'Treinamento'))

#4,5 - Calculo do Erro Relativo Medio e sua variancia
#IMPLEMENTE O CÓDIGO AQUI
E_ABS = np.abs(y_pred - vOutputsTrain)
ERM = np.average(E_ABS)
var_ERM = np.var(E_ABS)
print('Erro relativo medio do treinamento = {} - variancia = {}'.format(ERM, var_ERM))
#legenda
plt.gca().legend(('Previsão', 'Treinamento'))
plt.show()
```
